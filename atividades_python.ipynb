Escreva um programa que peça à pessoa usuária três números que representam os lados de um triângulo. O programa deve informar se os valores podem ser utilizados para formar um triângulo e, caso afirmativo, se ele é equilátero, isósceles ou escaleno. Tenha em mente algumas dicas:
- Três lados formam um triângulo quando a soma de quaisquer dois lados for maior que o terceiro;
- Triângulo Equilátero: três lados iguais;
- Triângulo Isósceles: quaisquer dois lados iguais;
- Triângulo Escaleno: três lados diferentes;


````python
print(f'Olá, este é nosso programa validador de triangulos.\n'
'Sabemos que triangulos são formados por 3 lados, certo?\n'
'Abaixo solicitaremos os números e veremos se estes valores podem formar um triangulo, lembrando que nenhum dos lados pode ser um valor negativo ou igual a 0.')

n1 = float(input('Digite o valor de um dos lados do triangulo: '))
n2 = float(input('Digite o outro lado: '))
n3 = float(input('Só mais um valor: '))

triangulo = n1 + n2 > n3 and n1 + n3 > n2 and n2 + n3 > n1

if n1 <= 0 or n2 <= 0 or n3 <= 0:
  print(f'\nNenhum dos lados pode ser menor ou igual a zero., tente novamente.')

elif not triangulo:
  print(f'\nEstes números não podem formar um triangulo :( ')

elif n1 == n2 == n3:
  print(f'\nEste triangulo é um triangulo equilátero!')

elif n1 != n2 and n1 != n3 and n2 != n3:
  print(f'\nEste é um triangulo Escaleno!')

elif n1 == n2 or n1 == n3 or n2 == n3:
  print(f'\nEste é um triangulo Isósceles!')

print(f'\nParabéns, o que achou do nosso programa?')
````

Escreva um programa que leia a quantidade de litros vendidos e o tipo de combustível (E para etanol e D para diesel) e calcule o valor a ser pago pelo cliente. Tenha em mente algumas dicas:

- O do valor do desconto será a multiplicação entre preço do litro, quantidade de litros e o valor do desconto.
- O valor a ser pago por um cliente será o resultado da multiplicação do preço do litro pela quantidade de litros menos o valor de desconto resultante do cálculo.

````python
print(f'Bem vindo ao posto de gasolina Venancio, aqui voce encontra os melhores preços!\n')

nome_cliente = input('Para iniciarmos o atendimento, qual é o seu nome? ')
tipo_combustivel = input('Pra começar, qual será o combustível que voce quer abastecer?\n\nResponda "D" para Diesel e "E" para Etanol. ')
litros = float(input('Muito bem! Agora, me fale, quantos litros voce precisa? '))

tipo_combustivel = tipo_combustivel.strip().upper()
nome_cliente = nome_cliente.strip().title()

if litros <= 0 or (tipo_combustivel != 'E' and tipo_combustivel != 'D'):
  print(f'O valor em litros ou tipo de combustível é inválido!')
else:
  if tipo_combustivel == 'E' and litros <= 15:
    valor_litro = 1.70
    valor_desconto = 0.98
  elif tipo_combustivel == 'E' and litros > 15:
    valor_litro = 1.70
    valor_desconto = 0.96
  elif tipo_combustivel == 'D' and litros <= 15:
    valor_litro = 2.00
    valor_desconto = 0.97
  else:
    valor_litro = 2.00
    valor_desconto = 0.95

  valor_total = ((valor_litro * litros) * valor_desconto)

  print(f'\nO valor a ser pago é: R${valor_total: .2f}'
      f'\n\nMuito obrigado por escolher o Posto Venancio Sr(a).{nome_cliente}, volte sempre!')
````

A partir do valor da variação, deve ser enviada às seguintes sugestões:
- Para variação acima de 20%: bonificação para o time de vendas.
- Para variação entre 2% e 20%: pequena bonificação para time de vendas.
- Para variação entre 2% e -10%: planejamento de políticas de incentivo às vendas.
- Para bonificações abaixo de -10%: corte de gastos.


````python
print(f'Bem vindo ao nosso planejamento estratégico, aqui avaliaremos as estratégias para o próximo ano de acordo com a variação nas vendas dos anos de 2022 e 2023.')

vendas_2022 = float(input('\nPra começarmos, digite a seguir a quantidade de vendas realizadas no ano de 2022: '))
vendas_2023 = float(input('\nÓtimo! Agora, digite a quantidade de vendas realizadas em 2023: '))


if vendas_2022 <= 0 or vendas_2023 <0:
  print(f'O valor de vendas foi inferior ou igual a 0, favor verificar!')
else:

  variacao = ((vendas_2023 - vendas_2022) / vendas_2022)

  if variacao > 0.20:
    estrategia = 'A equipe vendeu muito bem, a estratégia é promover bonificação!'
  elif variacao >= 0.02:
    estrategia = 'A equipe vendeu bem, mas poderia ter ido melhor, pequena bonificação.'
  elif variacao >= -0.10:
    estrategia = 'A equipe não foi bem, vamos fazer campanha para incentivo as vendas.'
  else:
    estrategia = 'Isso foi um desastre, precisamos cortar custos no próximo ano!!!'

  print(f'\nA variação foi de: {variacao:.1%}'
        f'\n\n{estrategia}')
````

````python
colonia_a = 4
colonia_b = 10
dias = 0

while colonia_a < colonia_b:
  colonia_b = colonia_b + (colonia_b * 0.015)
  colonia_a = colonia_a + (colonia_a * 0.03)
  dias = dias + 1

print(f'Quantidade de Bactéricas Colonia A: {colonia_a:.2f}'
f'\nQuantidade de Bactéricas Colonia B: {colonia_b:.2f}'
f'\nA colonia A demorou {dias} dias pra ultrapassar a Colonia B.')
````

#### Questão 4

````python
n = 0
celsius = 0
soma_temperaturas = 0
while celsius != -273:
  celsius = float(input('Digite a temperatura em graus Celcius: '))
  if celsius != -273:
    n += 1
    soma_temperaturas += celsius
if n > 0:
  media = soma_temperaturas / n
  print(f'\nA média das temperaturas digitadas foi de: {media:.2f}°C')
else:
  print('\nNenhuma temperatura válida foi digitada.')
````

#### Questão 5
Escreva um programa que calcule o fatorial de um número inteiro fornecido pela pessoa usuária. O fatorial de um número inteiro é a multiplicação desse número por todos os seus antecessores até o número 1. Por exemplo, o fatorial de 5 é 5 x 4 x 3 x 2 x 1 = 120.

````python
numero = int(input('Digite um número inteiro: '))
fatorial = 1

if numero < 0:
  print('Não existem fatoriais para números negativos!')
else:
  for i in range (numero,0,-1):
    fatorial *= i

  print(f'\nO número digitado foi: {numero} e o seu fatorial é: {fatorial}')
````
